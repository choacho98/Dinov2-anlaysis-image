```python
import torch
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from torchvision import transforms
from PIL import Image

# 1️⃣ 加载图片
image_path = "/content/arcade/arcade/id_30/id_30_139.578514_35.70389796_180.jpg"  # 替换成你的图片路径
img = Image.open(image_path).convert("RGB")

# 2️⃣ 预处理（和DINOv2模型输入一致）
transform = transforms.Compose([
    transforms.Resize((518, 518)),
    transforms.ToTensor(),
    transforms.Normalize(mean=(0.485, 0.456, 0.406),
                         std=(0.229, 0.224, 0.225))
])
input_tensor = transform(img).unsqueeze(0)

# 3️⃣ 载入DINOv2模型
dinov2 = torch.hub.load('facebookresearch/dinov2', 'dinov2_vits14')
with torch.no_grad():
    features = dinov2.get_intermediate_layers(input_tensor, n=1)[0]  # 提取最后层patch特征
    patch_tokens = features.squeeze(0)[1:, :]  # 去掉CLS token，得到 patch 向量
    n_patches = int(patch_tokens.shape[0] ** 0.5)
    print(f"每张图 {n_patches}×{n_patches} 个 patch")

# 4️⃣ PCA降维成3通道（R,G,B）
pca = PCA(n_components=3)
pca_features = pca.fit_transform(patch_tokens.cpu().numpy())
pca_features -= pca_features.min()
pca_features /= pca_features.max()

# 5️⃣ 自动推断 patch 网格大小
n_patches = patch_tokens.shape[0]
h = int(np.sqrt(n_patches))
w = int(np.ceil(n_patches / h))
print(f"Patch grid shape: {h}×{w} (共 {n_patches} 个 patch)")

# 6️⃣ 重塑成图片
pca_image = pca_features.reshape(h, w, 3)
plt.figure(figsize=(6,6))
plt.imshow(pca_image)
plt.axis('off')
plt.title("DINOv2 PCA 特征空间分布")
plt.show()

```

```python
from PIL import Image
import cv2

# 调整彩色特征图大小
pca_resized = cv2.resize((pca_image * 255).astype(np.uint8), img.size, interpolation=cv2.INTER_NEAREST)
overlay = cv2.addWeighted(np.array(img), 0.6, pca_resized, 0.4, 0)

plt.imshow(overlay)
plt.axis('off')
plt.title("DINOv2 特征叠加可视化")
plt.show()
```

```python
from PIL import Image, ImageOps
import torch
from torchvision import transforms
import os
import numpy as np

# 加载 DINOv2
model = torch.hub.load("facebookresearch/dinov2", "dinov2_vitb14")
model.eval()

# 预处理
transform = transforms.Compose([
    transforms.Resize((518, 518)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    ),
])

vectors = []
ids = []

root_dir = "/content/Silkroad/Silk road"   # ⭐ 你必须确认这个路径下有你的图片

valid_ext = (".jpg", ".jpeg", ".png", ".JPG", ".JPEG", ".PNG")

print("检测到的子文件夹：", os.listdir(root_dir))

for folder in sorted(os.listdir(root_dir)):
    sub_dir = os.path.join(root_dir, folder)
    if not os.path.isdir(sub_dir):
        continue

    print("\n正在处理目录：", folder)

    embeddings = []

    # 读取 4 个方向的照片
    for file in os.listdir(sub_dir):
        if file.endswith(valid_ext):
            img_path = os.path.join(sub_dir, file)

            try:
                img = Image.open(img_path)
                img = ImageOps.exif_transpose(img)   # ⭐ 自动处理手机旋转
                img = img.convert("RGB")

                img_t = transform(img).unsqueeze(0)

                with torch.no_grad():
                    feat = model(img_t)

                embeddings.append(feat.squeeze().numpy())
                print("  ✔ 成功读取：", file)

            except Exception as e:
                print("  ✘ 读取失败：", file, "错误：", e)

    if len(embeddings) == 0:
        print("  ⚠ 警告：该文件夹没有有效图片！跳过")
        continue

    # 每个点的四方向取平均
    mean_vec = np.mean(embeddings, axis=0)
    vectors.append(mean_vec)
    ids.append(folder)

vectors = np.array(vectors)
print("\n【最终结果】形态特征矩阵维度：", vectors.shape)
```

```python
import torch
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from torchvision import transforms
from PIL import Image
import cv2  # OpenCV 用于图像融合

# --- 配置 ---
# ⚠️ 请确保这里加载的模型与您提取特征用于聚类的模型一致！
# 如果您的CSV是768维，请使用 'dinov2_vitb14'
# 如果您的CSV是384维，请使用 'dinov2_vits14'
MODEL_TYPE = 'dinov2_vitb14'
IMAGE_PATH = r"/content/id_10_139.781072650028_35.7042970707355_0.jpg" # 替换为您的某张代表性图片路径

# 1️⃣ 加载图片
img_pil = Image.open(IMAGE_PATH).convert("RGB")
W, H = img_pil.size # 记录原图尺寸

# 2️⃣ 预处理（为了可视化，最好保持长宽比，这里简化为强制缩放以匹配patch）
# DINOv2 patch size = 14. 518 是 14 的倍数 (37 * 14)
target_size = (518, 518)
transform = transforms.Compose([
    transforms.Resize(target_size),
    transforms.ToTensor(),
    transforms.Normalize(mean=(0.485, 0.456, 0.406), std=(0.229, 0.224, 0.225))
])
input_tensor = transform(img_pil).unsqueeze(0)

# 3️⃣ 载入DINOv2模型
print(f"Loading {MODEL_TYPE}...")
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
dinov2 = torch.hub.load('facebookresearch/dinov2', MODEL_TYPE).to(device)
dinov2.eval()

# 4️⃣ 提取特征
with torch.no_grad():
    input_tensor = input_tensor.to(device)
    # 提取最后一层特征
    dict_feat = dinov2.forward_features(input_tensor)
    patch_tokens = dict_feat["x_norm_patchtokens"] # 直接获取 patch tokens

    # 自动计算 grid 大小
    B, N, C = patch_tokens.shape
    h_feat = int(np.sqrt(N))
    w_feat = int(np.sqrt(N))
    print(f"Feature grid: {h_feat}x{w_feat}, Dimension: {C}")

# 5️⃣ PCA 降维 (核心逻辑)
# 将 Tensor 转为 Numpy
features = patch_tokens[0].cpu().numpy() # Shape: (1369, 768)

# 主要是取前3个主成分作为RGB通道
pca = PCA(n_components=3)
pca_features = pca.fit_transform(features)

# 归一化到 0-255 以便绘图
pca_features = (pca_features - pca_features.min(0)) / (pca_features.max(0) - pca_features.min(0))
pca_image = (pca_features * 255).astype(np.uint8).reshape(h_feat, w_feat, 3)

# 6️⃣ 可视化增强：放大并叠加
# 将 PCA图 (37x37) 放大回 (518, 518)
pca_resized = cv2.resize(pca_image, target_size, interpolation=cv2.INTER_NEAREST)

# 读取原图用于叠加 (同样缩放到 518x518)
orig_img = np.array(img_pil.resize(target_size))

# 图像融合 (0.5 原图 + 0.5 PCA图)
overlay = cv2.addWeighted(orig_img, 0.5, pca_resized, 0.5, 0)

# 7️⃣ 绘图
plt.figure(figsize=(15, 5))

# 原图
plt.subplot(1, 3, 1)
plt.imshow(orig_img)
plt.title("Original Image")
plt.axis('off')

# PCA 特征图 (模型看到的语义)
plt.subplot(1, 3, 2)
plt.imshow(pca_resized)
plt.title("DINOv2 PCA Features (Semantic)")
plt.axis('off')

# 叠加图 (便于分析)
plt.subplot(1, 3, 3)
plt.imshow(overlay)
plt.title("Overlay Analysis")
plt.axis('off')

plt.tight_layout()
plt.show()
```
